### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
cleaned_df = df.copy()

cleaned_df["Released_Year"] = pd.to_numeric(cleaned_df["Released_Year"], errors="coerce")
cleaned_df["Runtime"] = pd.to_numeric(cleaned_df["Runtime"].astype(str).str.replace(" min", "", regex=False), errors="coerce")
cleaned_df["Gross"] = pd.to_numeric(cleaned_df["Gross"], errors="coerce")
cleaned_df["Meta_score"] = pd.to_numeric(cleaned_df["Meta_score"], errors="coerce")

for column in ["Released_Year", "Runtime", "Gross", "Meta_score"]:
    cleaned_df[column] = cleaned_df[column].fillna(cleaned_df[column].median())

for column in ["Certificate", "Genre", "Director", "Star1"]:
    cleaned_df[column] = cleaned_df[column].fillna("Unknown")

cleaned_df.to_csv("cleaned_movies.csv", index=False)

print("Cleaned dataset saved as cleaned_movies.csv")
print("Remaining missing values:")
print(cleaned_df.isnull().sum())
```

Har Director ki un movies ka count maloom karein jinhone 'No_of_Votes' 500,000 (5 lakh) se zyada hasil kiye hain.

```python
genre_gross = df[["Genre", "Gross"]].dropna(subset=["Genre"]).copy()
genre_gross["Genre"] = genre_gross["Genre"].str.split(", ")
genre_gross = genre_gross.explode("Genre")
genre_gross = genre_gross[genre_gross["Genre"].isin(["Drama", "Action"])]

drama_gross = genre_gross.loc[genre_gross["Genre"] == "Drama", "Gross"].dropna()
action_gross = genre_gross.loc[genre_gross["Genre"] == "Action", "Gross"].dropna()

plt.figure(figsize=(8, 5))
plt.boxplot([drama_gross, action_gross], tick_labels=["Drama", "Action"])
plt.title("Gross Revenue: Drama vs Action Movies")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")
plt.tight_layout()
plt.show()
```
